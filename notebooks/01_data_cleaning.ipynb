# 01 | Data Cleaning

Load the Kaggle video game sales CSV, validate its schema, normalize labels and numeric columns, impute missing release years and publishers, and save a clean dataset for downstream analysis.

The source file is expected at `data/raw/vgsales.csv`. Missing or unparseable years are median-imputed so those titles remain available for modeling; this means year-based analyses include imputed observations. Missing publishers are kept under an explicit `Unknown` category.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'src').is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

from src.data_cleaning import clean_sales_data, load_sales_data

RAW_PATH = PROJECT_ROOT / 'data' / 'raw' / 'vgsales.csv'
PROCESSED_DIR = PROJECT_ROOT / 'data' / 'processed'
CLEAN_PATH = PROCESSED_DIR / 'vgsales_clean.csv'
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
if not RAW_PATH.exists():
    raise FileNotFoundError(f'Place the Kaggle CSV at {RAW_PATH}')

raw = load_sales_data(RAW_PATH)
print(f'Loaded {len(raw):,} rows and {raw.shape[1]} columns')
raw.head()
```

## Clean and audit

Required game labels that are absent are dropped; duplicate rows are removed after normalization. Sales fields are numeric, with unparseable values treated as zero. The audit summarizes how much the source changed.

```python
cleaned, audit = clean_sales_data(raw)
display(pd.Series(audit, name='rows'))
cleaned.to_csv(CLEAN_PATH, index=False)
print(f'Saved {len(cleaned):,} cleaned rows to {CLEAN_PATH.relative_to(PROJECT_ROOT)}')
cleaned.head()
```
